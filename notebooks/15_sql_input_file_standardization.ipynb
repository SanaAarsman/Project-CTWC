```python
from pathlib import Path
import pandas as pd
```

```python
processed_path = Path("../data/processed")
csv_path = processed_path / "csv_files"
xlsx_path = processed_path / "xlsx_files"

csv_path.mkdir(parents=True, exist_ok=True)
xlsx_path.mkdir(parents=True, exist_ok=True)
```

```python
files_to_convert = [
    "ctwc_all_games_cleaned.xlsx",
    "ctwc_countries_cleaned.xlsx",
    "ctwc_player_info_cleaned.xlsx",
    "ctwc_year_stats_cleaned.xlsx"
]

for file_name in files_to_convert:
    input_path = processed_path / file_name
    output_name = file_name.replace(".xlsx", ".csv")
    output_path = csv_path / output_name
    
    df_temp = pd.read_excel(input_path)
    df_temp.to_csv(output_path, index=False)
    
    print(f"Converted: {file_name} -> {output_name}")
```
